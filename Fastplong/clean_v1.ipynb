```python
import os 
import subprocess
```

```python
from concurrent.futures import ThreadPoolExecutor
max_processes = 4

cpu_count = os.cpu_count() or 1
threads = max(2, cpu_count // 3)
```

```python
input_file = "/home/led/HLA_temp/Datas/hac_dir/"

input_files = [f for f in os.listdir(input_file) if f.endswith('.fastq.gz')]
```

```python
output_file = "/home/led/temp/development/fastplong/"

os.makedirs(output_file, exist_ok=True)
```

```python
sequence_formats = (
    ".fa", ".fasta", ".fas", ".fsa",
    ".fna", ".ffn", ".faa", ".frn",
    ".fq", ".fastq"
)

compression_formats = ("", ".gz", ".xz", ".zst", ".bz2", ".lz4")

file_formats = []

for sequence in sequence_formats:
    for compression in compression_formats:
        file_formats.append(sequence + compression)

input_files = [
    f for f in os.listdir(input_file)
    if f.lower().endswith(tuple(file_formats))
    and os.path.isfile(os.path.join(input_file, f))
]
```

```python
for i in input_files: 
    
    def run_fastplong(i):
    
        for suffix in file_formats:
            if i.lower().endswith(suffix):
                sample_id = i[:-len(suffix)]
                break
        
        input_path = os.path.join(input_file, i)
        
        output_file_final = os.path.join(output_file, sample_id)
        os.makedirs(output_file_final, exist_ok=True)
        
        output_fastq = os.path.join(
                output_file_final,
                f"{sample_id}_fastplong.fastq.gz",
            )
        output_html = os.path.join(
                output_file_final,
                f"{sample_id}_fastplong.html",
            )
        output_json = os.path.join(
                output_file_final,
                f"{sample_id}_fastplong.json",
            )
        
        cmd = ["fastplong", "-i", input_path, "-o", output_fastq, "--html", output_html, "--json", output_json]
        
        subprocess.run(cmd, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        
        return sample_id

    
with ThreadPoolExecutor(max_workers=max_processes) as executor:

    for sample_id in executor.map(run_fastplong, input_files):
        print(f"{sample_id} 过滤完成")
```
